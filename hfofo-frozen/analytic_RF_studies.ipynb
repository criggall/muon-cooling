```python
def dE_func(phi, phi_s, I=0):
    return np.sqrt(np.cos(phi_s)+np.cos(phi)+(phi-np.pi+phi_s)*np.sin(phi_s))
```

```python
phi_vals = np.linspace(-np.pi, np.pi, 1000)
```

```python
dE_vals = dE_func(phi_vals, 0)
```

```python
plt.plot(phi_vals, dE_vals, color='black')
plt.plot(phi_vals, -dE_vals, color='black')
plt.xlabel('$\phi$')
plt.ylabel('$\Delta E$');
```

```python
plt.plot(phi_vals, dE_func(phi_vals, 0), color='tab:blue', label='$\phi_s=0$')
plt.plot(phi_vals, -dE_func(phi_vals, 0), color='tab:blue')
plt.plot(phi_vals, dE_func(phi_vals, np.pi/16), color='tab:orange', label='$\phi_s=\pi/16$')
plt.plot(phi_vals, -dE_func(phi_vals, np.pi/16), color='tab:orange')
plt.plot(phi_vals, dE_func(phi_vals, np.pi/8), color='tab:green', label='$\phi_s=\pi/8$')
plt.plot(phi_vals, -dE_func(phi_vals, np.pi/8), color='tab:green')
plt.plot(phi_vals, dE_func(phi_vals, np.pi/4), color='tab:red', label='$\phi_s=\pi/4$')
plt.plot(phi_vals, -dE_func(phi_vals, np.pi/4), color='tab:red')
plt.plot(phi_vals, dE_func(phi_vals, np.pi/2), color='tab:purple', label='$\phi_s=\pi/2$')
plt.plot(phi_vals, -dE_func(phi_vals, np.pi/2), color='tab:purple')
plt.xlabel('$\phi$')
plt.ylabel('$\propto\Delta E$')
plt.legend(loc='upper left', fontsize=10);
```

```python
def dE_func(phi, phi_s, I=0):
    return np.sqrt(2*np.cos(phi_s)+2*np.cos(phi)+(phi-np.pi+phi_s)*np.sin(phi_s))
```
