## RF gradient 5% (1.25 MV/m), 10mm slab absorber

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

import sys
import os
parent_dir = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.append(parent_dir)

from functions.read_g4bl_data import readTraceData, readDetData
from functions.set_plot_settings import setPlotSettings

setPlotSettings(font=True)
```

### phaseAcc=0

```python
data = readTraceData('AllTracks_abstest_phAcc0.txt')
```

```python
# remove last period (for fringe-field mitigation)
# 2 periods HFOFO + 1 at start + 1 at end
period_len = 4200
data = data[data['z'] < 500+period_len*2]
```

```python
data = data.sort_values('z')
```

```python
plt.plot(data['x'], data['y'])
plt.xlabel('$x$ [mm]')
plt.ylabel('$y$ [mm]');
```

```python
plt.plot(data['z'], data['y'])
plt.xlabel('$z$ [mm]')
plt.ylabel('$y$ [mm]');
```

```python
plt.plot(data['z'], data['ptotal'],color='black')
plt.xlabel('$z$ [mm]')
plt.ylabel('$p_{total}$ [MeV/c]')
# plt.xlim(500,4700)
rf_len = 250
space = period_len/6
for i in range(12):
    z1 = 500-rf_len/2+space*i
    z2 = 500+rf_len/2+space*i
    plt.axvspan(xmin=z1-rf_len/2,xmax=z1+rf_len/2,color='tab:red',alpha=0.5,zorder=-0.5)
    plt.axvline(z2+(500-rf_len/2+space*(i+1)-z2)/2,color='darkcyan',alpha=0.5,zorder=-0.5)
    plt.axvspan(xmin=z2-rf_len/2,xmax=z2+rf_len/2,color='tab:red',alpha=0.5,zorder=-0.5)
```

```python
plt.plot(data['z'], data['ptotal'],color='black')
plt.xlabel('$z$ [mm]')
plt.ylabel('$p_{total}$ [MeV/c]')
plt.xlim(500,4700)
rf_len = 250
space = period_len/6
for i in range(12):
    z1 = 500-rf_len/2+space*i
    z2 = 500+rf_len/2+space*i
    plt.axvspan(xmin=z1-rf_len/2,xmax=z1+rf_len/2,color='tab:red',alpha=0.5,zorder=-0.5)
    plt.axvline(z2+(500-rf_len/2+space*(i+1)-z2)/2,color='darkcyan',alpha=0.5,zorder=-0.5)
    plt.axvspan(xmin=z2-rf_len/2,xmax=z2+rf_len/2,color='tab:red',alpha=0.5,zorder=-0.5)
```
