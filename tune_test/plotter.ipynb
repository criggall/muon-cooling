```python
import numpy as np
from matplotlib import pyplot as plt

import sys
import os
parent_dir = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.append(parent_dir)

from functions.read_g4bl_data import readTraceData
from functions.set_plot_settings import setPlotSettings

setPlotSettings(font=True)
```

## no RF, no absorbers

```python
refData = readTraceData('g4bl-output-noRF/ReferenceParticle.txt')
tuneData = readTraceData('g4bl-output-noRF/TuneParticle.txt')
```

```python
# remove extra solenoids for fringe-field mitigation:

refData = refData[refData['z'] > 500]
refData = refData[refData['z'] < 126275]

tuneData = tuneData[tuneData['z'] > 500]
tuneData = tuneData[tuneData['z'] < 126275]
```

```python
plt.plot(refData['x'],refData['y'],label='reference', color='black')
plt.plot(tuneData['x'],tuneData['y'],label='tune', color='tab:blue')
# plt.scatter(refData['x'][0], refData['y'][0], s=20, zorder=10, color='tab:green', label='start')
# plt.scatter(refData['x'][len(refData['x'])-1], refData['y'][len(refData['x'])-1], s=20, zorder=10, color='tab:red', label='end')
plt.xlabel('$x$ [mm]')
plt.ylabel('$y$ [mm]')
plt.legend(fontsize=12, loc='upper right');
```

```python
plt.plot(refData['z'],refData['ptotal'],label='reference', color='black')
plt.plot(tuneData['z'],tuneData['ptotal'],label='tune', color='tab:blue')
plt.xlabel('$z$ [mm]')
plt.ylabel('$p_{total}$ [MeV/c]')
plt.legend();
```

```python
plt.plot(refDataRF['z'],refDataRF['ptotal'],label='reference', color='black')
plt.scatter(tuneDataRF['z'],tuneDataRF['ptotal'],label='tune', color='tab:blue')
plt.xlabel('$z$ [mm]')
plt.ylabel('$p_{total}$ [MeV/c]')
plt.legend();
```

```python
plt.plot(refDataRF['z'],refDataRF['ptotal'],label='reference', color='black', lw=0.75)
plt.ylim((190,210))
plt.xlim((0,period_len+200))
# plt.scatter(tuneDataRF['z'],tuneDataRF['ptotal'],label='tune', color='tab:blue')
plt.xlabel('$z$ [mm]')
plt.ylabel('$p_{total}$ [MeV/c]')
plt.legend(fontsize=10)
for i in range(6):
    z1 = 500-rf_len/2+space*i
    z2 = 500+rf_len/2+space*i
    # plt.axvline(x=z1,color='tab:red',zorder=-1,lw=0.5)
    plt.axvspan(xmin=z1-rf_len/2,xmax=z1+rf_len/2,color='tab:red',alpha=0.5,zorder=-0.5)
    # plt.axvline(x=500+rf_len/2+space*i,color='tab:red',zorder=-1,lw=0.5)
    plt.axvspan(xmin=z2-rf_len/2,xmax=z2+rf_len/2,color='tab:red',alpha=0.5,zorder=-0.5)
```
